# 🛠️ Organização e Transformação de Dados com NumPy — GABARITO

**Contexto:** Preparação de bases e álgebra linear: sequências, tipagem, reshape, cópias, broadcasting, `argsort` e `linalg.solve`.

**Colab:** `Arquivo > Fazer upload do notebook`, executar com `Shift + Enter`. Só `numpy`.

```python
import numpy as np
```

## 🟢 Nível 1 — Geração de Sequências e Tipagem de Dados (Básico)

```python
# Passo 1: 30 dias do mês
dias_mes = np.arange(1, 31)
print("Dias do mês:", dias_mes)

# Passo 2: 5 metas igualmente espaçadas entre 20k e 30k
metas = np.linspace(20000, 30000, 5)
print("\nMetas semanais:", metas)

# Passo 3: estoque decimal → inteiro
estoque_decimal = np.array([10.5, 20.1, 30.9])
estoque_inteiro = estoque_decimal.astype(int)
print("\nEstoque (float):", estoque_decimal, "→ (int):", estoque_inteiro)

# Passo 4: últimos 5 dias, do mais recente para o mais antigo
ultimos_5_invertidos = dias_mes[-5:][::-1]  # também vale: dias_mes[:-6:-1]
print("\nÚltimos 5 dias invertidos:", ultimos_5_invertidos)
```

## 🟡 Nível 2 — Redimensionamento e Proteção de Dados (Intermediário)

```python
# Passo 1: 12 meses → matriz 4×3 (4 trimestres)
visitas = np.array([1200, 1400, 1100, 1500, 1300, 1600, 1450, 1700, 1550, 1800, 1650, 1900])
visitas_t = visitas.reshape(4, 3)
print("Matriz de trimestres (4×3):")
print(visitas_t)

# Passo 2: empilhar 1º semestre (marketing) + 2º semestre (comercial) verticalmente
sem1 = visitas_t[:2]   # 1º e 2º trimestres
sem2 = visitas_t[2:]   # 3º e 4º trimestres
anual = np.vstack([sem1, sem2])
print("\nEmpilhado via np.vstack (reconstrói 4×3):")
print(anual)

# Passo 3: fatiar 1º trimestre com cópia (protege o original)
projecao = visitas_t[0].copy()
projecao[0] = 0
print("\nProjeção (copy) alterada:", projecao)
print("Original intacto?", visitas_t[0], "← sim, copia protegida")

# Passo 4: achatar a matriz de volta em 1D
visitas_1d = visitas_t.ravel()
print("\nMatriz achatada (ravel):", visitas_1d)
```

## 🔵 Nível 3 — Ranking, Broadcasting e Resolução de Sistemas (Avançado)

```python
# Passo 1: ranking via np.argsort
pontuacoes = np.array([85, 92, 78, 95, 88])
indices_ranking = np.argsort(pontuacoes)
print("Pontuações dos alunos:", pontuacoes)
print("Índices da ordenação (do menor p/ maior):", indices_ranking)
print("Notas já ordenadas:", pontuacoes[indices_ranking])

# Passo 2: preços em formato de linha → coluna via np.newaxis
precos = np.array([100, 200, 300])
precos_coluna = precos[:, np.newaxis]
print("\nPreços linha:", precos, "shape:", precos.shape)
print("Preços coluna:\n", precos_coluna, "shape:", precos_coluna.shape)

# Broadcasting: cada preço × descontos
descontos = np.array([0.9, 0.8, 0.7])
print("\nPreços × descontos (broadcasting):\n", precos_coluna * descontos)

# Passo 3: sistema linear — custo de placas e sensores
# Projeto A: 2 placas + 1 sensor = 500; Projeto B: 1 placa - 1 sensor = 100
A = np.array([[2, 1], [1, -1]])
b = np.array([500, 100])
custo = np.linalg.solve(A, b)
print(f"\nCusto placa robótica: R$ {custo[0]:.2f}")
print(f"Custo sensor: R$ {custo[1]:.2f}")
```

## 🎓 Resumo

```python
print("=" * 55)
print("RESUMO NUMPY — ORGANIZAÇÃO")
print("=" * 55)
print(f"Meses/trimestres: {visitas_t.shape} | ravel → {visitas_1d.shape}")
print(f"Melhor aluno: posição {indices_ranking[-1]} com {pontuacoes[indices_ranking[-1]]}")
print(f"Sistema: placa=R$ {custo[0]:.2f}, sensor=R$ {custo[1]:.2f} (verificação: 2*{custo[0]:.0f}+{custo[1]:.0f}={2*custo[0]+custo[1]:.0f})")
print("=" * 55)
```
