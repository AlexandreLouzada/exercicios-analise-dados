## 🧱 Célula 1: Configuração do Ambiente e Criação dos Dados

▶️ Execute primeiro — gera o dataset com imperfeições simuladas (nulos e duplicatas).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configuração visual do Seaborn
sns.set_theme(style="whitegrid")
np.random.seed(101)

# Gerando dados fictícios de clientes de um e-commerce
n = 200
categorias = ['Eletrônicos', 'Vestuário', 'Casa e Decoração', 'Livros']

dados = {
    'ID_Cliente': np.random.randint(1000, 1050, size=n),
    'Categoria': np.random.choice(categorias, size=n),
    'Valor_Gasto': np.random.normal(loc=250, scale=80, size=n).round(2),
    'Avaliacao_Compra': np.random.choice([1, 2, 3, 4, 5, np.nan], size=n, p=[0.1, 0.1, 0.2, 0.3, 0.2, 0.1]),
    'Idade_Cliente': np.random.randint(18, 65, size=n)
}

df_ecommerce = pd.DataFrame(dados)

# Inserindo algumas linhas duplicadas intencionalmente
df_ecommerce = pd.concat([df_ecommerce, df_ecommerce.iloc[:10]], ignore_index=True)

print("Dataset gerado com sucesso!")
print(f"Total de linhas no dataset: {len(df_ecommerce)}")
df_ecommerce.head()
```

## ✅ Tarefa 1: Limpeza e Preparação dos Dados

**Pergunta de negócio:** A base está confiável para análise ou tem duplicatas e avaliações faltantes?

1. Verifique quantas linhas duplicadas existem e remova-as.
2. Identifique quais colunas possuem valores nulos (`NaN`).
3. Preencha os nulos de `Avaliacao_Compra` com a mediana da coluna.

```python
# 1. Verificar duplicadas
n_dup = df_ecommerce.duplicated().sum()
print(f"Linhas duplicadas encontradas: {n_dup}")

# Remover duplicadas (mantém a primeira ocorrência)
df_ecommerce = df_ecommerce.drop_duplicates(ignore_index=True)
print(f"Total após remover duplicatas: {len(df_ecommerce)}")

# 2. Identificar colunas com valores nulos
print("\nValores nulos por coluna:")
print(df_ecommerce.isnull().sum())

# 3. Preencher nulos de Avaliacao_Compra com a mediana
mediana = df_ecommerce["Avaliacao_Compra"].median()
print(f"\nMediana de Avaliacao_Compra: {mediana}")
df_ecommerce["Avaliacao_Compra"] = df_ecommerce["Avaliacao_Compra"].fillna(mediana)

print("\nNulos após preenchimento:")
print(df_ecommerce.isnull().sum())
df_ecommerce.head()
```

💡 **Por que a mediana?** Avaliação é escala ordinal (1–5) com possíveis outliers. A mediana não é distorcida por extremos, ao contrário da média. Alternativa: `mode()` se quiser o valor mais frequente.

## ✅ Tarefa 2: Comparação por Categoria (Boxplot com Seaborn)

1. Crie um `sns.boxplot` comparando `Valor_Gasto` entre as `Categoria`.
2. Identifique qual categoria tem maior variação ou mais outliers.

```python
# Estatísticas de apoio à leitura do boxplot
print(df_ecommerce.groupby("Categoria")["Valor_Gasto"].describe().round(2))
```

```python
# Boxplot: distribuição de Valor_Gasto por Categoria
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_ecommerce, x="Categoria", y="Valor_Gasto", palette="Set2")
plt.title("Distribuição do Valor Gasto por Categoria", fontsize=14, fontweight="bold")
plt.xlabel("Categoria", fontsize=12)
plt.ylabel("Valor Gasto (R$)", fontsize=12)
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()

# 💡 Leitura: compare a altura da caixa (IQR = variação central) e os pontos fora dos "bigodes" (outliers).
# Com seed=101, rode a célula de estatísticas acima para apontar objetivamente a categoria de maior desvio-padrão.
```

## ✅ Tarefa 3: Distribuição e Correlação

1. Crie um histograma com curva de densidade (`sns.histplot` com `kde=True`) para `Idade_Cliente`.
2. Adicione título e rótulos adequados aos eixos.

```python
# Histograma + KDE da idade dos clientes
plt.figure(figsize=(10, 6))
sns.histplot(data=df_ecommerce, x="Idade_Cliente", bins=15, kde=True, color="steelblue")
plt.title("Distribuição da Idade dos Clientes", fontsize=14, fontweight="bold")
plt.xlabel("Idade do Cliente (anos)", fontsize=12)
plt.ylabel("Frequência", fontsize=12)
plt.tight_layout()
plt.show()

print(f"Idade média: {df_ecommerce['Idade_Cliente'].mean():.1f} | Mediana: {df_ecommerce['Idade_Cliente'].median():.0f}")
```

## 🎓 Conclusão — Resumo executivo


```python
print("=" * 60)
print("RELATÓRIO — E-commerce (base limpa)")
print("=" * 60)
print(f"Registros finais: {len(df_ecommerce)}")
print(f"Avaliação média (após preencher mediana): {df_ecommerce['Avaliacao_Compra'].mean():.2f}")
print("\nGasto médio por categoria:")
print(df_ecommerce.groupby('Categoria')['Valor_Gasto'].mean().round(2).sort_values(ascending=False))
print("\nVariação (desvio-padrão) por categoria:")
print(df_ecommerce.groupby('Categoria')['Valor_Gasto'].std().round(2).sort_values(ascending=False))
print("=" * 60)
```
