# 🏪 Análise de Dados no Varejo com NumPy — GABARITO

**Contexto:** Equipe de dados de rede de lojas. Insights rápidos com NumPy: vetorização, máscaras, NaN, `np.where`, categorias e destaques de marketing.

**Colab:** `Arquivo > Fazer upload do notebook`, executar com `Shift + Enter`. Só `numpy`.

---

```python
import numpy as np
```

## 🟢 Nível 1 — Faturamento Semanal e Filtros (Básico)

```python
# Passo 1: array 1D do histórico diário
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])
print("Pecas vendidas:", pecas_vendidas)

# Passo 2: faturamento = vetorização (array * 50)
faturamento_semana = pecas_vendidas * 50
print("Faturamento diário (R$):", faturamento_semana)
print(f"Faturamento total da semana: R$ {faturamento_semana.sum():,.2f}")

# Passo 3: máscara booleana — dias com mais de 200 peças
mascara_pico = pecas_vendidas > 200
print("\nMáscara (dias > 200):", mascara_pico)
print("Vendas só nos dias de pico:", pecas_vendidas[mascara_pico])
print(f"Quantidade de dias de pico: {mascara_pico.sum()}")
```

## 🟡 Nível 2 — Múltiplas Filiais e Falhas de Sistema (Intermediário)

```python
# Passo 1: matriz 2D — lojas nas linhas, dias nas colunas
vendas_filiais = np.array([
    [200, 220, np.nan, 250],  # Loja A (falha no 3º dia)
    [150, 180, 160, 190],     # Loja B
    [300, 310, 290, 330]      # Loja C
])
print("Matriz de vendas (lojas × dias):")
print(vendas_filiais)

# Passo 2: total por loja (soma nas linhas)
total_por_loja = np.nansum(vendas_filiais, axis=1)
print("\nTotal de peças por loja:", total_por_loja)

# Passo 3: média diária geral ignorando NaN + média contaminada p/ comparação
media_nanmean = np.nanmean(vendas_filiais)
media_contaminada = np.mean(vendas_filiais)
print(f"\nMédia diária geral (ignorando NaN) = {media_nanmean:.2f}")
print(f"Média contaminada (np.mean)        = {media_contaminada:.2f}  ← não usar")

# Passo 4: meta diária de 200 → novo array condicional
status_meta = np.where(vendas_filiais >= 200, 'Meta Atingida', 'Abaixo')
print("\nStatus da meta por loja/dia:")
print(status_meta)
```

## 🔵 Nível 3 — Categorização e Destaques de Marketing (Avançado)

```python
# Passo 1: gerador moderno com semente fixa + amostra de 50 vendas
rng = np.random.default_rng(42)
vendas_departamento = rng.choice(['Eletrônicos', 'Roupas', 'Casa'], size=50)
print("Amostra:", vendas_departamento[:10], "...")

# Passo 2: contagem por departamento (np.unique + return_counts)
departamentos, contagens = np.unique(vendas_departamento, return_counts=True)
print("\nVolume de vendas por departamento:")
for dept, qtd in zip(departamentos, contagens):
    print(f"  {dept}: {qtd} vendas")

# Passo 3: campanha campeã via índice do maior valor
faturamento_campanhas = np.array([12000, 45000, 23000, 89000, 31000])
indice_campea = np.argmax(faturamento_campanhas)
print(f"\nCampanha campeã: posição {indice_campea} com R$ {faturamento_campanhas[indice_campea]:,.2f}")
```

## 🎓 Resumo executivo

```python
print("=" * 62)
print("RELATÓRIO VAREJO — NumPy")
print("=" * 62)
print(f"Faturamento semanal: R$ {faturamento_semana.sum():,.2f} | picos: {mascara_pico.sum()} dias")
print(f"Maior loja: Loja {np.argmax(total_por_loja) + 1} com {int(total_por_loja.max())} peças")
print(f"Média diária geral (sem NaN): {media_nanmean:.2f}")
print(f"Departamento líder: {departamentos[np.argmax(contagens)]} ({contagens.max()} de 50 vendas)")
print(f"Campanha destaque: índice {indice_campea} — R$ {faturamento_campanhas[indice_campea]:,.2f}")
print("=" * 62)
```
